### class Dog — 普通 Python 类
```python
  class Dog:
      species = "犬科动物"

      def __init__(self, name, age):
          self.name = name
          self.age = age
```
  - 手动编写 __init__，手动赋值 self.xxx = xxx
  - 没有数据验证——传入 Dog("旺财", "三岁") 或 Dog(123, None) 都不会报错
  - 属性可以随意增删——dog1.foo = "随便加" 完全合法
  - 就是一个普通的 Python 对象

### class APIResponse(BaseModel) — Pydantic 模型
```python
  class APIResponse(BaseModel):
      code: int = 0
      message: str = "success"
      data: Any = None
```
  - 不需要写 __init__，Pydantic 自动生成
  - 自动类型验证——传入 APIResponse(code="abc") 会直接报错，因为 code 声明了 int
  - 自动类型转换——传入 APIResponse(code="123") 会自动把字符串 "123" 转成整数 123
  - 不可随意增删属性——只能有声明过的字段
  - 自带 .model_dump() 序列化为 dict、.model_dump_json() 序列化为 JSON 字符串
  - FastAPI 能自动识别 BaseModel，生成 OpenAPI 文档、做请求/响应校验
